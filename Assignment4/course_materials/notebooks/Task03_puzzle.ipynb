Spanish: Importamos example3.rdf en nuestro grafo

English: Import example3.rdf in our graph

```python
from rdflib import Graph, Namespace, Literal
g = Graph()

#TODO - Complete
g.parse(github_storage+"/rdf/example3.rdf", format="xml")
```

Spanish: Listar todos los recursos con la propiedad VCARD:FN

English: List all resources with the VCARD:FN property

```python
VCARD = Namespace("http://www.w3.org/2001/vcard-rdf/3.0#")

for s,p,o in g.triples((None, VCARD.FN, None)):
  print(s)
```

Spanish: Listar el nombre completo (VCARD:FN) de estos recursos

English: List the full name (VCARD:FN) of these resources

```python
for s,p,o in g.triples((None, VCARD.FN, None)):
  print(o)
```

Spanish: Listar ahora todos los recursos que tienen el nombre de familia "Smith", debemos tener en cuenta el tipo de dato de nuestro literales a la hora de realizar búsquedas. En nuestro ejemplo los literales están tipados como XSD String (http://www.w3.org/2001/XMLSchema#string) y al buscar debemos indicarlo, de lo contrario la comparación fallará aunque las cadenas de texto sean iguales.

English: Llist all resources with the family name "Smith", we must take into account the data type of our literals when performing searches. In our example, the literals are typed as XSD String (http://www.w3.org/2001/XMLSchema#string) and we must specify this when searching; otherwise, the comparison will fail even if the text strings are identical.

```python
from rdflib import XSD
for t in g.triples((None, VCARD.Family, Literal('Smith', datatype=XSD.string))):
  print(t)
```

Spanish: Comprobar comparaciones entre literales

English: Check comparisons between literals

```python
print(Literal(1, datatype=XSD.number).eq(Literal("1", datatype=XSD.string)))
print(Literal(1, datatype=XSD.number).eq(Literal(1, datatype=XSD.string)))
print(Literal(1).eq(Literal(1, datatype=XSD.number)))
print(Literal(1, datatype=XSD.number).eq(Literal(1, datatype=XSD.number)))
print(Literal(1).datatype)
print(Literal("1").datatype)
```

Spanish: Listar todos los recursos que contienen un email

English: List all resources that contains an email

```python
#from rdflib import FOAF
FOAF = Namespace("http://xmlns.com/foaf/0.1/")
for s,p,o in g.triples((None, FOAF.email, None)):
  print(s,p,o)
```

Spanish: Listar todos los sujetos que conocen a "Jane Smith" (VCARD:FN) por su nombre de pila (VCARD:Given). Cuando sepamos que vamos a obtener un solo resultado podemos emplear la función value indicando cada parámetro conocido (solo uno debe ser None).

English: List all subjects who know "Jane Smith" (VCARD:FN) by her first name (VCARD:Given). When we know that we will obtain a single result, we can use the "value" function, specifying each known parameter (only one of which must be "None").

```python
janeSmithURI = g.value(subject=None, predicate=VCARD.FN, object=Literal("Jane Smith"))
knows = g.triples((None, FOAF.knows, janeSmithURI))

for s,p,o in knows:
  givenName = g.value(subject=s, predicate=VCARD.Given, object=None)
  print(givenName)
```
